```python
import numpy as np
```

```python
class NeuralNetwork:

    def __init__(self, input_size, hidden_size, output_size, learning_rate):
        rng = np.random.default_rng()

        self.W1 = rng.standard_normal(size=(input_size, hidden_size)) * 0.5
        self.b1 = np.zeros((1, hidden_size))

        self.W2 = rng.standard_normal(size=(hidden_size, output_size)) * 0.5
        self.b2 = np.zeros((1, output_size))

        self.lr = learning_rate


    def relu(self, z):
        return np.maximum(0, z)

    def relu_derivative(self, z):
        return (z > 0).astype(int)         

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-np.clip(z, -500, 500)))


    def forward(self, X):
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = self.relu(self.z1) 

        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = self.sigmoid(self.z2)

        return self.a2


    def backward(self, X, y, y_hat):
        m = X.shape[0] 

        # Output layer gradients
        dz2 = y_hat - y                                   # dL/dz2 = dL/dŷ . dŷ/dz2
        dW2 = (1/m) * np.dot(self.a1.T, dz2)              # dL/dW2 = dL/dW2 . dz2/dW2 
        db2 = (1/m) * np.sum(dz2, axis=0, keepdims=True)  # dL/db2 = dl/dz2 . dz2/db2

        # Hidden layer gradients  
        dz1 = np.dot(dz2, self.W2.T) * self.relu_derivative(self.z1) # dL/dz1 = dL/dŷ . dŷ/dz2 . dz2/da1 . da1/dz1
        dW1 = (1/m) * np.dot(X.T, dz1)                               # dL/dW2 = dL/dz1 . dz1/dW1
        db1 = (1/m) * np.sum(dz1, axis=0, keepdims=True)             # dL/db1 = dL/dz1 . dz1/db1

        # Update parameters
        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1


    def train(self, X, y, epochs):
        for epoch in range(epochs):
            y_hat = self.forward(X)
            self.backward(X, y, y_hat)  

            if epoch % 50 == 0:
                loss = np.mean((y_hat - y)**2)
                print(f"Epoch {epoch:5d} | loss: {loss:6f}")


    def predict(self, X):
        output = self.forward(X)
        return (output > 0.5).astype(int)
```

```python
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

y = np.array([[0],
              [1],
              [1],
              [0]])

nn = NeuralNetwork(2, 4, 1, 0.1)
nn.train(X, y, epochs=1000)

nn.predict(X)
```
